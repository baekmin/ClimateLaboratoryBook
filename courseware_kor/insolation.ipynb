## 1. 일사량(Insolation) 분포

:::{note}
이 노트는 Hartmann의 『지구 물리 기후학』 2.7절을 면밀히 따르고 있습니다.

대기권 상단에 도달하는 **태양 복사량**(우리가 '일사량(Insolation)'이라고 부르는 것)은 다음 요인에 따라 달라집니다.

- 위도
- 계절
- 하루 중 시간

이 일사량은 기후 시스템의 주요 원동력입니다. 여기서는 일사량을 결정하는 기하학적 요인들을 살펴볼 것이며, 주로 **일 평균(daily average)** 값에 초점을 맞출 것입니다.

### 태양 천정각

우리는 **태양 천정각 (solar zenith angle)** $\theta_s$를 지구 표면의 특정 지점에서의 법선과 지구 표면의 한 지점과 태양을 잇는 선 사이의 각으로 정의합니다.

![하트만 그림 2.5](https://github.com/baekmin/ClimateLaboratoryBook/blob/main/images/Hartmann_Fig2.5.png?raw=1)

위 그림({cite:t}`Hartmann`에서 재인용)에서 그림자 면적 대 표면 면적의 비율은 태양 천정각(Solar Zenith Angle)의 코사인 값과 같습니다.

### 순간 태양 플럭스

단위 면적당 태양 플럭스는 다음과 같이 쓸 수 있습니다.

$$ Q = S_0 \left( \frac{\overline{d}}{d} \right)^2 \cos \theta_s $$

여기서 $\overline{d}$는 플럭스 밀도 $S_0$(즉, 태양 상수)가 측정되는 평균 거리이며, $d$는 태양으로부터의 실제 거리입니다.

- 어떤 상황에서 이 비율이 항상 1이 될까요?

### 천정각 계산

플럭스(flux) 자체와 마찬가지로, 태양 천정각(Solar Zenith Angle)은 위도, 계절, 그리고 하루 중 시간에 따라 달라집니다.

#### 적위각 (Declination angle)
계절 의존성은 태양의 **적위각** (declination angle)으로 표현될 수 있습니다. 적위각은 정오에 태양 바로 아래에 있는 지구 표면 지점의 위도를 나타내며, $\delta$로 표시됩니다.

$\delta$는 현재 북반구 하지점(6월 21일)의 +23.45º에서 북반구 동지점(12월 21일)의 -23.45º 사이로 변화합니다.

#### 시각

**시각(Hour angle)** $h$는 태양직하점의 정오 위치에 대한 상대적인 경도로 정의됩니다.

#### 천정각 공식
이러한 정의들과 구형 기하학(spherical geometry)을 활용하여 (하트만(Hartmann) 저서의 부록 A 참조), 우리는 모든 위도 $\phi$, 계절, 그리고 하루 중 시간에 대한 태양 천정각(solar zenith angle)을 다음과 같이 표현할 수 있습니다.

$$ \cos \theta_s = \sin \phi \sin \delta + \cos\phi \cos\delta \cos h $$

#### 일출과 일몰

만약 $\cos\theta_s < 0$이면 태양은 지평선 아래에 있으며 일사량(Insolation)은 0입니다 (즉, 밤입니다!).

일출과 일몰은 태양 천정각(solar zenith angle)이 90º이고 따라서 $\cos\theta_s=0$일 때 발생합니다. 위의 공식은 다음과 같이 주어집니다.

$$ \cos h_0 = - \tan\phi \tan\delta $$

여기서 $h_0$는 일출과 일몰 시의 시각각(hour angle)입니다.

#### 극야

극지방 근처에서는 특별한 조건들이 나타납니다. $\phi$와 $\delta$의 부호가 같을 때 여름철에는 $90º-\delta$보다 극쪽에 있는 위도들은 지속적으로 햇빛을 받습니다. 극점에서는 6개월간 백야(perpetual daylight)가 지속되며, 이 기간 동안 태양은 지평선 위 일정한 각도 $\delta$를 유지하며 방위를 따라 움직입니다.

겨울철에는 $\phi$와 $\delta$의 부호가 반대이며, $90º-|\delta|$보다 극쪽에 있는 위도들은 영구적인 어둠(perpetual darkness) 속에 잠깁니다. 극점에서는 6개월의 낮이 6개월의 낮과 교대합니다.

적도에서는 낮과 밤이 일년 내내 각각 12시간씩 지속됩니다.

### 일평균 일사량

태양 천정각(solar zenith angle) 표현식을 일사량(insolation) 공식에 대입하면 위도, 계절, 시간에 따른 순간 일사량은 다음과 같습니다:

$$ Q = S_0 \left( \frac{\overline{d}}{d} \right)^2 \Big( \sin \phi \sin \delta + \cos\phi \cos\delta \cos h  \Big) $$

이 공식은 낮 시간 동안($|h| < h_0$)에만 유효하며, 그 외의 경우(밤)에는 $Q=0$입니다.

일일 평균 일사량(daily average insolation)을 얻기 위해, 우리는 이 식을 일출부터 일몰까지 적분하고 24시간(또는 일일 시간을 시각(hour angle)으로 표현하므로 $2\pi$ 라디안)으로 나눕니다.

$$ \overline{Q}^{day} = \frac{1}{2\pi} \int_{-h_0}^{h_0} Q ~dh$$

$$ = \frac{S_0}{2\pi} \left( \frac{\overline{d}}{d} \right)^2 \int_{-h_0}^{h_0} \Big( \sin \phi \sin \delta + \cos\phi \cos\delta \cos h  \Big) ~ dh $$

이는 쉽게 적분하여 일일 평균 일사량(Daily Average Insolation)에 대한 공식을 얻을 수 있습니다:

$$ \overline{Q}^{day} = \frac{S_0}{\pi} \left( \frac{\overline{d}}{d} \right)^2 \Big( h_0 \sin\phi \sin\delta + \cos\phi \cos\delta \sin h_0 \Big)$$

여기서 일출/일몰 시의 시각(時角) (Hour Angle) $h_0$는 라디안(Radians) 단위여야 합니다.

### 일일 평균 천정각

지구 표면(특히 수역)의 광학적 특성 때문에 표면 알베도(surface albedo)가 태양 천정각에 따라 달라지는 것으로 나타났습니다. 따라서 일광 시간 동안의 평균 태양 천정각을 위도와 계절의 함수로 고려하는 것이 유용합니다.

여기서 적절한 일일 평균은 시간에 따른 가중치가 아닌, 일사량(insolation)에 따라 가중된 값입니다. 그 공식은 다음과 같습니다.

$$ \overline{\cos\theta_s}^{day} = \frac{\int_{-h_0}^{h_0} Q \cos\theta_s~dh}{\int_{-h_0}^{h_0} Q ~dh} $$

{cite:t}`Cronin:2014a`는 또한 일사량으로 가중된 값 $\overline{\cos\theta_s}^{day}$이 다양한 상황에서 산란 편향을 최소화하는 데 좋은 선택임을 보여줍니다.

$\overline{\cos\theta_s}^{day}$를 계산하기 위해, 우선 위 식의 분자와 분모에서 인자 $S_0 \left(\frac{\overline{d}}{d}\right)^2$가 상쇄되어 다음이 남는다는 점을 주목합니다.

$$ \overline{\cos\theta_s}^{Q} = \frac{\int_{-h_0}^{h_0} \cos^2 \theta_s ~dh}{\int_{-h_0}^{h_0} \cos\theta_s ~dh} $$

약간의 계산 과정을 거치면, 위도 $\phi$, 적위(declination angle) $\delta$, 그리고 시간각(hour angle) $h$로 나타낸 $\cos\theta_s$에 대한 식을 대입하여 위 식의 분자와 분모를 적분할 수 있습니다. 저는 그 결과로 도출된 공식을 아래 Python 코드에 구현했습니다.

````python
import numpy as np
import matplotlib.pyplot as plt
from numpy import sin,cos,tan,arccos,arcsin,arctan,pi,abs

lat = np.linspace(-90., 90, 500)
phi = np.deg2rad(lat)

declinations = {'21 December': -23.45,
                'Equinox': 0.,
                '21 June': 23.45}

def hour_angle_at_sunset(delta, phi):
    return np.where( abs(delta)-pi/2+abs(phi) < 0., # there is sunset/sunrise
              arccos(-tan(phi)*tan(delta)),
              # otherwise figure out if it's all night or all day
              np.where(phi*delta>0., pi, 0.) )

def insolation_weighted_coszen(delta, phi):
    h0 = hour_angle_at_sunset(delta, phi)
    denominator = h0*sin(phi)*sin(delta) + cos(phi)*cos(delta)*sin(h0)
    numerator = (h0*(2* sin(phi)**2*sin(delta)**2 + cos(phi)**2*cos(delta)**2) +
        cos(phi)*cos(delta)*sin(h0)*(4*sin(phi)*sin(delta) + cos(phi)*cos(delta)*cos(h0)))
    return numerator / denominator / 2

for label, delta in declinations.items():
    plt.plot(lat, np.rad2deg(arccos(insolation_weighted_coszen(np.deg2rad(delta), phi))), label=label)
plt.legend()
plt.grid()
plt.title('Insolation-weighted daily average solar zenith angle as a function of latitude');
plt.ylim(30,90);
plt.xlim(-90,90);
````

_위 그림은 {cite:p}`Hartmann`의 그림 2.8을 재현한 것입니다._

평균 천정각은 열대지방보다 극지방에서 훨씬 더 높습니다. 이는 고위도에서 관측되는 매우 높은 표면 반사율(알베도)의 한 원인이 됩니다.

### 실제 달력 날짜 사용

`daily_insolation` 함수에 대한 `day` 입력값은 특정 달력 연도와 연동되지 않으며, 실제로는 춘분(80일로 정의됨)으로부터의 시간 길이(소수일 단위)를 측정합니다.

실제 달력 연도는 길이가 가변적입니다(예: 윤년).

다행히도 Python에서 날짜와 시간을 관리하는 데 도움이 되는 사용하기 쉬운 몇 가지 소프트웨어 패키지가 있습니다. 여기서는 `np.datetime64` 및 `np.timedelta64` 데이터 타입을 통해 numpy의 일부 기능을 사용할 것입니다.

먼저 실제 달력 날짜/시간을 태양 계산기에 필요한 일(day) 인덱스로 변환하는 도우미 함수(helper function)를 정의합니다.

````python
def dates_to_day_index(datetime):
    '''Convert dates and time (assumed to be UTC) to a fractional number of days
    from a hypothetical January 1 00h that is exactly 80 days before the spring equinox.

    Input datetime is a np.datetime64 object (or array of these)

    Output is array of floats.
    '''
    # Precise time of the most recent Spring Equinox in 9:01 UTC on March 20 2025
    spring_equinox = np.datetime64('2025-03-20T09:01')
    time_since_equinox = datetime - spring_equinox
    time_since_jan1 = time_since_equinox + np.timedelta64(80, 'D')
    return time_since_jan1 / np.timedelta64(1, 'D')  # result as fractional days
````

이제 우리는 특정 날짜를 나타내는 날짜/시간 객체(datetime object)를 정의하고, 모든 위도에서의 일사량(Insolation)을 그래프로 나타낼 것입니다.

````python
# Here we generate a datetime64 object with a specific date:
Oct15 = np.datetime64('2025-10-15')
Oct15
````

````python
Q = daily_insolation(lat, dates_to_day_index(Oct15))
fig, ax = plt.subplots()
ax.plot(lat,Q)
ax.set_xlim(-90,90); ax.set_xticks([-90,-60,-30,-0,30,60,90])
ax.set_xlabel('Latitude')
ax.set_ylabel('W/m2')
ax.grid()
ax.set_title('Daily average insolation at {}'.format(Oct15))
````

## 3. 일사량의 전 지구적, 계절별 분포 (현재의 궤도 매개변수)

현재의 궤도 매개변수(orbital parameters)를 사용하여 1년 및 모든 위도에 대한 일사량(Insolation) 배열을 계산합니다. 멋진 등고선도(contour plot)를 만들기 위해 조밀한 격자를 사용할 것입니다.

````python
lat = np.linspace( -90., 90., 500)
days = np.arange(0., 1., 0.001) * const.days_per_year

Q = daily_insolation( lat, days )
````

그리고 Q를 위도와 연중 시간의 함수로 등고선 플롯을 만드세요.

````python
fig, ax = plt.subplots(figsize=(10,8))
CS = ax.contour( days, lat, Q , levels = np.arange(0., 600., 50.) )
ax.clabel(CS, CS.levels, inline=True, fmt='%1.0f', fontsize=10)
ax.set_xlabel('Days since January 1', fontsize=16 )
ax.set_ylabel('Latitude', fontsize=16 )
ax.set_title('Daily average insolation', fontsize=24 )
ax.contourf ( days, lat, Q, levels=[-1000., 0.], colors='k' )
````

### 시간 및 공간 평균

Q의 면적 가중 전 지구 연평균을 구합니다.

````python
Qaverage = np.average(np.mean(Q, axis=1), weights=np.cos(np.deg2rad(lat)))
print( 'The annual, global average insolation is %.2f W/m2.' %Qaverage)
````

또한 일 년 중 몇몇 다른 시점의 위도대별 평균 일사량(Insolation)을 플로팅하세요:

````python
summer_solstice = 472  # index of the solstice along this time axis
winter_solstice = 976  # and same for winter solstice
fig, ax = plt.subplots(figsize=(10,8))
ax.plot( lat, Q[:,(summer_solstice, winter_solstice)] );
ax.plot( lat, np.mean(Q, axis=1), linewidth=2 )
ax.set_xbound(-90, 90)
ax.set_xticks( range(-90,100,30) )
ax.set_xlabel('Latitude', fontsize=16 );
ax.set_ylabel('Insolation (W m$^{-2}$)', fontsize=16 );
ax.grid()
````
