## 1. 단일 층 대기

우리는 가장 간단한 온실 모델, 즉 장파 복사(longwave radiation)를 흡수하고 방출할 수 있는 단일 층 대기를 사용하여 온실 효과를 정량화하는 첫 시도를 할 것입니다.

![단일층 대기 모식도](https://github.com/baekmin/ClimateLaboratoryBook/blob/main/images/1layerAtm_sketch.png?raw=1)

### 가정

- 대기는 온도 $T_a$의 단일 공기층이다.
- 대기는 **단파 복사(shortwave radiation)**에 **완전히 투명하다**.
- **지표면**은 단파 복사 $(1-\alpha) Q$를 흡수한다.
- 대기는 **적외 복사(infrared radiation)**에 **완전히 불투명하다**.
- 지표면과 대기 모두 **흑체(blackbody)**로서 복사를 방출한다 ($\sigma T_s^4, \sigma T_a^4$).
- 대기는 **상하 방향으로 동일하게** 복사를 방출한다 ($\sigma T_a^4$).
- 다른 열 전달 메커니즘은 없다.

이제 우리는 에너지 균형(energy balance) 개념을 사용하여 지표면과 대기에서 에너지 수지(energy budgets)의 균형을 맞추기 위해 필요한 온도가 얼마인지, 즉 **복사 평형 온도(radiative equilibrium temperatures)**가 무엇인지 알아볼 수 있다.

### 지표면 에너지 평형

$$
\begin{align*}
\text{energy in} &= \text{energy out} \\
(1-\alpha) Q + \sigma T_a^4 &= \sigma T_s^4 \\
\end{align*}
$$

상층 대기가 존재한다는 것은 추가적인 에너지원(source term)이 있음을 의미합니다: 대기로부터의 하향 장파 복사(downwelling infrared radiation).

우리는 이를 **대기 역복사(back radiation)**라고 부릅니다.

### 대기권의 에너지 균형

$$
\begin{align}
\text{energy in} &= \text{energy out} \\
\sigma T_s^4 &= A\uparrow + A\downarrow = 2 \sigma T_a^4 \\
\end{align}
$$

이는 다음을 의미합니다.

$$ T_s = 2^\frac{1}{4} T_a \approx 1.2 T_a $$

따라서 우리는 순수한 **복사 평형(Radiative Equilibrium)**을 가지기 위해서는 $T_s > T_a$ 이어야만 한다는 것을 방금 확인했습니다.

*지표면은 대기보다 따뜻해야 합니다.*

### 복사 평형 표면 온도 구하기

이제 이것을 표면 방정식에 대입하여 다음을 얻습니다:

$$ \frac{1}{2} \sigma T_s^4 = (1-\alpha) Q $$

그리고 방출 온도($T_e$)의 정의를 사용하여 다음과 같이 쓸 수 있습니다:

$$ (1-\alpha) Q = \sigma T_e^4 $$

*사실, 이 모델에서 $T_e$는 대기 온도($T_a$)와 동일합니다. 이는 모든 외향 장파 복사(OLR)가 이 층에서 비롯되기 때문입니다.*

지표면 온도(surface temperature)를 구하면 다음과 같습니다:

$$ T_s = 2^\frac{1}{4} T_e $$

관측된 수치인 $T_e = 255$ K를 대입하면, 지표면 온도(surface temperature)는 다음과 같습니다:

$$T_s = 303 ~\text{K}$$

이 모델은 현실에 한 걸음 더 가까워진 것입니다: 지표면(surface)은 대기(atmosphere)보다 따뜻하고, 우주로의 방출은 대기 중에서 발생하며, 대기(atmosphere)는 아래에서부터 가열되어 지표면(surface)을 따뜻하게 유지하는 데 도움을 줍니다.

### 이 모델은 왜 지표면 온도(surface temperature)를 과대 예측할까요?

우리 모델은 현재 지표면 온도를 약 15ºC(관측된 288 K와 비교하여 303 K)만큼 과대 예측하고 있습니다.

그 이유에 대해 어떤 아이디어가 있으신가요?

기본적으로 위에 제시된 우리의 **가정 목록**을 읽어보면, 그 어떤 가정도 아주 좋은 근사치가 아니라는 것을 알 수 있습니다:

- 대기는 일부 태양 복사(solar radiation)를 흡수합니다.
- 대기는 장파 복사(longwave radiation)의 완벽한 흡수체가 아닙니다.
- 흡수와 방출은 파장에 따라 크게 달라집니다 *(대기는 흑체(blackbody)처럼 행동하지 않습니다)*.
- 방출은 단일 온도 $T_a$에 의해 결정되는 것이 아니라, 기온의 상세한 *수직 분포(vertical profile)*에 의해 결정됩니다.
- 에너지는 다양한 역학적 수송 메커니즘(예: 대류(convection) 및 경계층 난류(boundary layer turbulence))에 의해 수직 방향으로 재분배됩니다.

## 2. 2층 회색 기체 모델 소개

위 모델을 조금만 더 일반화하여, 조금 더 현실적인 장파 복사 전달(longwave radiative transfer) 모델을 구축해 봅시다.

우리는 단일층 모델이 가진 두 가지 단점을 다룰 것입니다.
1. 연직 구조(vertical structure)의 부재
2. 100% 장파 불투명도(longwave opacity)

이 두 가지 가정을 완화함으로써, **온실 효과(greenhouse effect)가 어떻게 작동하는지 이해하는 데** 매우 유용한 프로토타입 모델을 얻을 수 있습니다.

### 가정

- 대기는 **단파 복사(shortwave radiation)에 투명하다** (여전히)
- 대기를 **동일한 질량의 두 층**으로 나눈다 (따라서 구분선은 500 hPa 기압 고도에 있다)
- 각 층은 입사하는 장파 복사(longwave radiation) 중 **오직 $\epsilon$의 비율만을 흡수한다.**
- 우리는 비율 $\epsilon$을 층의 **흡수율(absorptivity)**이라고 부를 것이다.
- 각 층에서 $\epsilon$이 동일하다고 가정한다.

이것은 **회색 기체(grey gas)** 모델이라고 불리는데, 여기서 **회색(grey)**은 방출과 흡수가 **스펙트럼 의존성이 없다** (모든 파장에서 동일하다)는 것을 의미한다.

우리는 이 모델을 비공식적으로 "새는 온실(leaky greenhouse)"이라고 생각할 수 있다.

$\epsilon$이 각 층에서 동일하다는 가정은 흡수가 실제로 대기 중에 **잘 혼합된** 기체에 의해 이루어질 경우에 적절하다는 점에 유의하십시오.

우리의 두 가지 가장 중요한 흡수체(absorber) 중에서:

- CO$_2$는 잘 혼합되어 있습니다.
- H$_2$O는 그렇지 않습니다 (포화 증기압(saturation vapor pressure)의 강한 온도 의존성 때문에 주로 대류권(troposphere) 하층에 국한됩니다).

하지만 우리는 현재로서는 이러한 현실적 측면을 무시하겠습니다.

### 키르히호프의 법칙

우리의 모델을 구축하기 위해, 우리는 **키르히호프의 법칙**으로 알려진 추가적인 물리 원리를 도입해야 합니다.

$$ \text{absorptivity} = \text{emissivity} $$

따라서 **온도 $T$를 가진 대기층이** 입사하는 장파 복사(longwave radiation)의 **$\epsilon$ 비율을 흡수한다면**, 그 대기층은 반드시

$$ \epsilon ~\sigma ~T^4 $$

를 위쪽과 아래쪽 모두로 **방출해야 합니다**.

### 2층 대기에서의 복사 플럭스 개요

![2층 대기 개요도](https://github.com/baekmin/ClimateLaboratoryBook/blob/main/images/2layerAtm_sketch_general.png?raw=1)

- 표면 온도는 $T_s$입니다.
- 대기 온도는 $T_0, T_1$이며, 이때 $T_0$는 표면에 가장 가까운 층의 온도입니다.
- 각 대기층은 흡수율(absorptivity) $\epsilon_0$ 또는 $\epsilon_1$을 가집니다.
- 표면 방출(emission)은 $\sigma T_s^4$입니다 (즉, 표면은 흑체(blackbody)처럼 복사를 방출합니다).
- 대기 방출(emission)은 $\epsilon_0 \sigma T_0^4, \epsilon_1 \sigma T_1^4$입니다 (상방 및 하방으로).
- 각 대기층에 대해 흡수율(Absorptivity) = 방출률(Emissivity)이 성립합니다.
- 장파 복사(longwave radiation) 중 흡수되지 않는 비율 $(1-\epsilon_0)$ 또는 $(1-\epsilon_1)$은 각 층을 통해 **투과**됩니다.

## 3. 장파 복사의 상향 광선 추적

상향 장파 복사(upwelling longwave radiation)를 $U$라고 표기해봅시다.

### 지표에서 0층까지

지표에서 시작해봅시다. **지표에서 0층으로**의 상향 플럭스는 다음과 같습니다.

$$U_0 = \sigma T_s^4$$

(이는 지표에서 단순히 방출되는 복사입니다.)

### 0층에서 1층으로

이제 **이 복사속을 위쪽으로 따라가면서**, 우리는 먼저 이 복사속의 $\epsilon_0$만큼의 비율이 0층에서 **흡수**됨을 알 수 있습니다.

0층에서 1층으로 향하는 상향 복사속은 두 부분의 합으로 구성됩니다:

1.  아래에서 입사하는 복사속 중 **투과된 부분** (즉, **흡수되지 않은 부분**)
2.  0층에서 **새로 방출되는 상향 복사**

우리는 0층에서 1층으로 가는 이 상향 복사속을 다음과 같이 쓸 수 있습니다:

$$U_1 = (1-\epsilon_0) \sigma T_s^4 + \epsilon_0 \sigma T_0^4$$

### 1층을 넘어서

동일한 복사선을 계속 추적하면서, 우리는 동일한 논리를 적용합니다! $U_1$의 $\epsilon_1$ 비율에 해당하는 양은 1층에서 흡수되며, 따라서 투과되는 부분은 $(1-\epsilon_1) U_1$ 입니다.

1층에서 발생하는 새로운 방출(emissions)을 포함하면, 1층 위쪽으로 향하는 상향 복사 플럭스(upwelling flux)는 다음과 같습니다:

$$U_2 = (1-\epsilon_1) U_1 + \epsilon_1 \sigma T_1^4$$

### 외향 장파 복사 (Outgoing Longwave Radiation)

**1층 위에는 더 이상 대기가 없으므로**, 이 상향 복사선은 이 모델에서의 우리의 외향 장파 복사(OLR)입니다:

$$\text{OLR} = U_2 = (1-\epsilon_1) U_1 + \epsilon_1 \sigma T_1^4$$

여기서 위의 $U_1$ 표현식을 대입하면 다음과 같이 정리됩니다.

$$\text{OLR} = (1-\epsilon_1)(1-\epsilon_0) \sigma T_s^4 + (1-\epsilon_1)\epsilon_0\sigma T_0^4 + \epsilon_1 \sigma T_1^4$$

여기서 세 항은 **세 가지 층 각각에서 발생하는 총 외향 장파 복사(OLR)에 대한 기여**를 나타냅니다.

이 식에서 매개변수 $\epsilon_0, \epsilon_1$이 층의 *방출률*(emissivity)과 복사선이 통과하는 층의 *흡수율*(absorptivity)을 모두 나타내며 어떻게 다양한 방식으로 사용되는지 주목하십시오.

### 흡수율/방출률이 크거나 작을 때의 한계

다음 두 가지 질문에 대해 생각해 봅시다:

- 만약 $\epsilon_0 = \epsilon_1 =1$ 이면, 이 표현식은 어떻게 될까요? *이것은 물리적으로 무엇을 나타낼까요?*
- $\epsilon_0 = \epsilon_1 =0$ 인 경우는 어떨까요?

:::{important}
대기가 다른 층에서 방출되는 복사를 부분적으로 흡수하게 됨으로써, 우리는 이제 우주로 방출되는 장파 복사(Outgoing Longwave Radiation)가 **모든 층에서 방출되는 복사를 포함하며** - 따라서 **모든 층의 온도에 의해 영향을 받음**을 알 수 있습니다!
:::

### 두 층에서 동일한 방출률이라는 단순화 가정

우리는 **두 개의 새로운 매개변수**인 방출률(emissivities, 또는 흡수율, 같은 개념입니다) $\epsilon_0, \epsilon_1$을 도입했습니다. 따라서 지금까지 우리의 방정식은 두 층의 매우 다양한 대기 조성(예: 서로 다른 흡수 기체 또는 구름 양)을 설명하기에 충분히 일반적입니다.

단순화를 위해, 지금부터는 **두 대기층이 동일한 방출률(emissivity)을 가진다고 가정**할 것입니다. 즉, 다음과 같이 가정합니다.

$$ \epsilon_0 = \epsilon_1 = \epsilon $$

이러한 가정으로, OLR(장파 복사 방출량)에 대한 식은 다음으로 간소화됩니다.

$$\text{OLR} = (1-\epsilon)^2 \sigma T_s^4 + \epsilon(1-\epsilon)\sigma T_0^4 + \epsilon \sigma T_1^4$$

이것이 바로 다음 섹션에서 우리 모델을 관측값에 맞춰 조정하는 데 사용할 식입니다.

```python



# Install metpy if not already installed
import subprocess
import sys

try:
    from metpy.plots import SkewT
except ImportError:
    print("Installing metpy...")
    subprocess.check_call([sys.executable, "-m", "pip", "install", "metpy"])
    print("Installation complete. Please rerun the cell.")
    from metpy.plots import SkewT



#  This code is used just to create the skew-T plot of global, annual mean air temperature
import numpy as np
import matplotlib.pyplot as plt
import xarray as xr
from metpy.plots import SkewT

ncep_url = "http://www.esrl.noaa.gov/psd/thredds/dodsC/Datasets/ncep.reanalysis.derived/"
time_coder = xr.coders.CFDatetimeCoder(use_cftime=True)
ncep_air = xr.open_dataset( ncep_url + "pressure/air.mon.1981-2010.ltm.nc",
                           decode_times=time_coder)

#  Take global, annual average and convert to Kelvin
coslat = np.cos(np.deg2rad(ncep_air.lat))
weight = coslat / coslat.mean(dim='lat')
Tglobal = (ncep_air.air * weight).mean(dim=('lat','lon','time'))

fig = plt.figure(figsize=(9, 9))
skew = SkewT(fig, rotation=30)
skew.plot(Tglobal.level, Tglobal, color='black', linestyle='-', linewidth=2, label='Observations')
skew.ax.set_ylim(1050, 10)
skew.ax.set_xlim(-75, 45)
# Add the relevant special lines
skew.plot_dry_adiabats(linewidth=0.5)
skew.plot_moist_adiabats(linewidth=0.5)
skew.ax.legend()
skew.ax.set_title('Global, annual mean sounding from NCEP Reanalysis', fontsize = 16);
```

### 흡수체 추가에 따른 방출 수준의 변화

추가적인 **온실 흡수체(greenhouse absorbers)**를 더하면, 입사하는 장파 복사(longwave radiation)의 **더 큰 부분**이 **각 층에서 흡수**될 것입니다.

따라서 온실가스를 추가할수록 **$\epsilon$ 값은 증가**해야 합니다.

초기 상태에서 $\epsilon$을 가지고 있으며, 흡수율(absorptivity)이 $\epsilon_2 = \epsilon + \Delta \epsilon$으로 증가한다고 가정해 봅시다.

또한, 이러한 증가가 **갑작스럽게** 발생하여 온도가 이 변화에 반응할 시간이 없다고 가정해 봅시다. 우리는 대기 기둥(column) 내의 온도를 **고정된 상태로 유지**하고, 복사 플럭스(radiative fluxes)가 어떻게 변하는지 질문합니다.

**질문: 외향 장파 복사(Outgoing Longwave Radiation, OLR)가 증가할 것으로 예상하십니까, 아니면 감소할 것으로 예상하십니까?**

### 배출 수준 변화 계산

이 질문에 대한 답을 알아보기 위해 우리의 2층 누출(leaky) 온실 모델을 활용해 보겠습니다.

섭동 전 OLR (외향 장파 복사)의 구성 요소들은 다음과 같습니다.

$$
\begin{align*}
\text{OLR}_s &= (1-\epsilon)^2 \sigma T_s^4  \\
\text{OLR}_0 &= \epsilon(1-\epsilon)\sigma T_0^4 \\
\text{OLR}_1 &= \epsilon \sigma T_1^4
\end{align*}
$$

그리고 교란(perturbation) 후에는 다음을 얻습니다.

$$
\begin{align*}
\text{OLR}_s &= (1-\epsilon - \Delta \epsilon)^2 \sigma T_s^4  \\
\text{OLR}_0 &= (\epsilon + \Delta \epsilon)(1-\epsilon - \Delta \epsilon)\sigma T_0^4 \\
\text{OLR}_1 &= (\epsilon + \Delta \epsilon) \sigma T_1^4
\end{align*}
$$

각 층으로부터의 **외향 장파 복사(OLR) 변화**에 대한 기여도를 얻기 위해 원래 구성 요소들을 빼보겠습니다.

$$
\begin{align*}
\Delta \text{OLR}_s &= \left[(1-\epsilon - \Delta \epsilon)^2 - (1-\epsilon)^2\right]\sigma T_s^4  \\
\Delta \text{OLR}_0 &= \left[(\epsilon + \Delta \epsilon)(1-\epsilon - \Delta \epsilon) - \epsilon(1-\epsilon)  \right] \sigma T_0^4 \\
\Delta \text{OLR}_1 &= \left[(\epsilon + \Delta \epsilon) - \epsilon \right] \sigma T_1^4
\end{align*}
$$

이제 이를 전개해 봅시다. 하지만 다루기 쉽게 하기 위해 $\Delta \epsilon^2$ 항은 무시합니다 (이 항은 매우 작으며, 우리는 엡실론($\epsilon$)의 10% 미만 변화를 고려할 것입니다):

$$
\begin{align*}
\Delta \text{OLR}_s &\approx (\Delta \epsilon) \left[ -2(1-\epsilon) \right] \sigma T_s^4  \\
\Delta \text{OLR}_0 &\approx (\Delta \epsilon) (1 - 2 \epsilon) \sigma T_0^4 \\
\Delta \text{OLR}_1 &\approx (\Delta \epsilon) \sigma T_1^4
\end{align*}
$$

이제 각 항의 **부호**를 살펴보세요. $0 < \epsilon < 1$임을 기억하세요. **OLR(외향 장파 복사량)에서 어떤 항들이 증가하고 어떤 항들이 감소하는가요?**

**이것은 매우 중요하니, 잠시 멈춰서 생각해 보세요.**

**표면(surface)**으로부터의 기여는 **감소**해야 하고, 반면에 **상층(top layer)**으로부터의 기여는 **증가**해야 합니다.

**흡수체(absorbers)를 추가하면, 평균 방출(emission) 수준이 올라갑니다!**

## 6. 2층 회색 기체 모델에서의 복사 강제력 (Radiative Forcing)

### 복사 강제력 (Radiative Forcing)의 정의

이제 매우 중요한 양을 정의합니다:

**"복사 강제력 (Radiative Forcing)"은 흡수체를 추가한 후 TOA(대기 상단)에서 발생하는 총 복사 플럭스의 변화입니다.**

이 모델에서는 **장파 복사속(longwave flux)만 변할 수 있으므로**, 복사 강제력(Radiative Forcing)을 다음과 같이 계산합니다.

$$ R = - \Delta \text{OLR} $$

(음의 부호를 사용하는 이유는 기후 시스템이 **추가 에너지를 얻을 때 $R$이 양수가 되도록** 하기 위함입니다.)

### 복사 강제력과 복사 방출 고도 간의 연관성

우리는 방금, 우리가 **추가적인 흡수체**를 더할 때마다 (평균적으로) **우주로의 복사 방출**이 대기 중 **더 높은 고도**에서 발생할 것이라는 것을 알아냈습니다.

이것이 외향 장파 복사(OLR: Outgoing Longwave Radiation)에 어떤 의미일까요? 그것은 증가할까요, 아니면 감소할까요?

답을 얻기 위해, 우리는 위에 언급했던 세 가지 기여도를 합산해야 합니다:

$$
\begin{align*}
R &= -\Delta \text{OLR}_s - \Delta \text{OLR}_0 - \Delta \text{OLR}_1  \\
  &= -\Delta \epsilon \left[  -2(1-\epsilon) \sigma T_s^4 + (1 - 2 \epsilon) \sigma T_0^4  + \sigma T_1^4 \right]
\end{align*}
$$

이것은 양수입니까, 음수입니까? 핵심은 다음과 같습니다:

**온도, 즉 기온 감률(lapse rate)에 따라 달라집니다.**

### 등온 대기에서의 온실 효과

잠시 멈춰서 다음 질문을 고민해봅시다:

**지표면과 대기가 모두 동일한 온도에 있다면**, $\epsilon$이 증가할 때 (즉, 흡수 물질이 더 많아질 때) OLR(외향 장파 복사)은 증가할까요, 아니면 감소할까요?

이 질문에 대한 이해는 온실 효과의 작동 원리를 파악하는 데 매우 중요합니다.

#### 등온 경우를 풀어봅시다

우리는 복사 강제력(Radiative Forcing)에 대한 위 식에서 $T_s = T_0 = T_1$로 간단히 설정할 것입니다.

어떤 결과를 얻을 수 있을까요?

#### 답은 $R=0$ 입니다.

등온 대기의 경우, 추가 온실 흡수체(greenhouse absorbers)를 추가해도 장파 복사 에너지 방출량(OLR)에는 **변화가 없습니다**. 따라서 복사 강제력(Radiative Forcing)이 없으며 온실 효과(greenhouse effect)도 없습니다.

왜 그럴까요?

복사 방출 고도는 여전히 높아져야 합니다. 하지만 상층 고도(upper level)의 온도가 다른 모든 곳과 **같기** 때문에, 방출량은 정확히 동일합니다.

### 복사 강제력 (외향 장파 복사량(OLR) 변화)은 기온 감률에 따라 달라집니다!

온실 흡수체(greenhouse absorbers) 증가로 인한 복사 강제력(radiative forcing)의 더 현실적인 예시를 위해, 우리는 관측된 온도와 $\epsilon$에 대한 조정된 값을 사용합니다.

$\epsilon$가 2% 증가할 때, 우리는 답을 W m$^{-2}$ 단위로 나타낼 것입니다:

$$ \Delta \epsilon = 0.02 \times 0.58 $$

```python
epsilon = 0.586041150248834
delta_epsilon = 0.02 * epsilon
delta_epsilon
```

복사 강제력(Radiative Forcing)의 세 가지 구성 요소를 계산하세요:

```python
sigma = 5.67E-8
Ts = 288.
T0 = 275.
T1 = 230.
```

```python
#  Component originating from the surface
Rs = -delta_epsilon * (-2*(1-epsilon)*sigma * Ts**4)
Rs
```

```python
#  Component originating from level 0
R0 = -delta_epsilon * (1-2*epsilon) * sigma * T0**4
R0
```

```python
#  Component originating from level 1
R1 = -delta_epsilon * sigma * T1**4
R1
```

그러므로 총 복사 강제력(Radiative Forcing)을 얻기 위해 그것들을 합산하면 됩니다:

```python
R = Rs + R0 + R1
R
```

따라서, 우리의 예시에서 **외향 장파 복사(OLR)는 2.6 W m$^{-2}$만큼 감소**하며, 이는 **복사 강제력이 +2.6 W m$^{-2}$**와 동등합니다.

:::important} 우리가 방금 무엇을 계산했나요?

**관측된 기온 감률(lapse rates)을 고려할 때, 흡수체(absorbers)의 약간의 증가는 외향 장파 복사(OLR)를 약간 감소시킬 것입니다.**

따라서 **온실 효과는 더 강해지고**, 시스템에 에너지가 축적되기 시작할 것입니다. 이는 시스템이 새로운 평형 상태에 맞춰 조절됨에 따라 궁극적으로 **기온 상승을 야기**할 것입니다.

- 지표면 위에 **장파 흡수층**을 놓으면, 대기로부터의 **하향 복사(backradiation)** 때문에 **지표면**이 **상당히 더 따뜻하게** 유지됩니다 (온실 효과).

- **불완전 흡수** ($\epsilon < 1$)가 있을 경우, 모든 고도와 지표면에서 외향 장파 복사(OLR)에 기여하게 됩니다 (단일한 **방출 고도**는 존재하지 않습니다).
- 흡수 물질(absorber)을 더 추가하면, **상층 대기**에서의 **외향 장파 복사(OLR) 기여**는 **증가**하는 반면, 지표면으로부터의 기여는 **감소**합니다.
- 여러 고도에서의 가중치(weighting)가 상층으로 이동하는 이러한 현상이 바로 우리가 **방출 고도**가 **상승**한다고 말할 때의 의미입니다.

- 흡수체의 증가로 인한 **방사 강제력**(Radiative Forcing)은 **기온 감률**(Lapse Rate)에 따라 달라집니다.
- **등온 대기**(Isothermal Atmosphere)에서는 방사 강제력이 0이며, **온실 효과**(Greenhouse Effect)가 발생하지 않습니다.
- 우리 대기에서는 **대류권 온도**(Tropospheric Temperature)가 **고도에 따라 감소하는 경향이 있기 때문에** 방사 강제력은 양의 값을 가집니다.
